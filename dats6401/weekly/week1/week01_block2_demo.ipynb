## Part 2 — A dataset, a chart, its anatomy (~12 min)

`plotly.express` ships small teaching datasets — no download, no file path, no excuses.

````python
import plotly.express as px

df = px.data.tips()
print(df.shape)
df.head()
````

````python
# Tidy check, narrated: one row = one party (the observation),
# one column = one variable. This is the contract every chart assumes.
df.dtypes
````

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.scatter(df["total_bill"], df["tip"], alpha=0.6, color="#2E6E8E")
ax.set_xlabel("Total bill ($)")          # label WITH units
ax.set_ylabel("Tip ($)")
ax.set_title("Tips rise with the bill — and so does the spread")
plt.show()
````

**Narrate the anatomy as you type it:** figure → axes → marks → labels → title. Then the design question for the room: *the title states a finding, not "Figure 1" — why does that matter for a reader with 10 seconds?*

## Part 3 — The four-line app (~10 min)

Same data, same chart — now an application. Create `app.py` next to this notebook:

```python
import streamlit as st
import plotly.express as px

st.title("My First App")
df = px.data.tips()
st.dataframe(df)
st.scatter_chart(df, x="total_bill", y="tip")
```

Then in a terminal:

```bash
streamlit run app.py
```

**Demonstrate:** edit the title while it runs → hot reload. Resize the browser → the chart adapts. *How* this works (the re-run model) is Week 5; today the point is the barrier height: four lines.
